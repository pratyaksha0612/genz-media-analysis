```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("sentiment_labeled_data.csv")

# Percentage distribution
phase_sentiment = df.groupby(["phase", "sentiment"]).size().unstack()
phase_sentiment_percent = phase_sentiment.div(phase_sentiment.sum(axis=1), axis=0) * 100

print("Sentiment Percentage by Phase:\n")
print(phase_sentiment_percent)

# Plot
phase_sentiment_percent.plot(kind="bar", figsize=(8,6))
plt.ylabel("Percentage")
plt.title("Sentiment Distribution Across Phases")
plt.xticks(rotation=0)
plt.show()
```
